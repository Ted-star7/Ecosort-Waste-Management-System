## Part 2 — Transfer-learning CNN (MobileNetV2)
Train, tune, evaluate with confusion matrix + class-imbalance handling.

```
from src import cnn_model
train_ds,val_ds,test_ds,names = data_prep.make_image_datasets()
cw = cnn_model.compute_class_weights(train_ds)
cw
```

```
model,hist = cnn_model.train(train_ds,val_ds,epochs=12,fine_tune_epochs=5,class_weight=cw)
```

```
acc,cm,rep = cnn_model.evaluate(model,test_ds)
print('Test accuracy:',acc)
print(rep)
```

```
import seaborn as sns, matplotlib.pyplot as plt
sns.heatmap(cm,annot=True,fmt='d',xticklabels=C.CATEGORIES,yticklabels=C.CATEGORIES)
plt.xticks(rotation=45,ha='right'); plt.title('CNN confusion matrix'); plt.tight_layout(); plt.show()
```
